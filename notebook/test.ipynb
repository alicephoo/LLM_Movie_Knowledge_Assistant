```python
query = "science fiction movie about space travel"

candidates = hybrid_search_movies(
    query,
    limit=20
)

print("Number of candidates:", len(candidates))
```

```python
for rank, movie in enumerate(candidates, start=1):
    print(
        f"{rank}. {movie['title']} "
        f"(RRF={movie['score']:.6f})"
    )
```

```python
reranked_results = rerank_movies(
    query,
    candidates,
    limit=5
)
```

```python
for rank, movie in enumerate(
    reranked_results,
    start=1
):
    print(
        f"{rank}. {movie['title']} "
        f"(rerank={movie['rerank_score']:.4f})"
    )
```
